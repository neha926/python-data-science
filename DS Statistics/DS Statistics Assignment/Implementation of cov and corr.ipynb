```python
Question 9: Your team has collected customer satisfaction survey data on a scale of 1-10 and wants to understand its distribution before launching a new product. 
●	Explain which summary statistics and visualizations (e.g. mean, standard deviation, histogram) you’d use. 
●	Write Python code to create a histogram using Matplotlib for the survey data: 
```

```python
✔ Summary Statistics:

Mean (average): Customers kitna satisfied है, overall idea मिलता है.

Median: Middle value बताती है, agar data skewed हो तो ज़्यादा useful.

Mode: सबसे common satisfaction score.

Standard Deviation: Customers ki satisfaction kitni vary हो रही है.

Minimum–Maximum: Lowest aur highest satisfaction.

✔ Visualizations:

Histogram: Score ka distribution दिखाता है (kitne लोग 7,8,9 दे रहे हैं).

Boxplot: Outliers aur spread समझने के लिए.

Bar chart (optional): Agar tum frequency देखना चाहो.

Short summary line:
“Mean, median, mode se central tendency pata चलेगा, standard deviation se variability, aur histogram & boxplot se distribution clearly dikhega.”
```

```python
survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7] 
```

```python
survey_scores
```

```python
# mean : how much customer satisfied, we got overall idea 
import numpy as np 
np.mean(survey_scores)
```

```python
# median : if data is skewed its useful.
np.median(survey_scores) 
```

```python
# mode : most common satisfaction score
import statistics
statistics.mode(survey_scores)
```

```python
# std : how much customer satisfaction vary
np.std(survey_scores)
```

```python
# min,max : to know the satisfaction
min(survey_scores)
```

```python
max(survey_scores)
```

```python
import matplotlib.pyplot as plt
plt.hist(survey_scores,bins=10, edgecolor='black')
plt.xlabel("Customer Satisfaction Score (1-10)")
plt.ylabel("Frequency")
plt.title("Distribution of Customer Satisfaction Survey")
plt.show()
```

```python
# boxplot : if there is outlier(<1 and >10) present to analyze
sns.boxplot(survey_scores)
```
