<h4>Range</h4>

```python
data=[2,3,4,5,8,9,12,40,41,15]
```

```python
data
```

```python
max(data)-min(data)
```

<h4>Percentile</h4>

```python
import numpy as np 
q=np.percentile(data,[0,25,50,75,100])
```

```python
q
```

<h4>IQR</h4>

```python
iqr=q[3]-q[1]
iqr
```

```python
uf=q[3]+1.5*iqr
uf
```

```python
lf=q[3]-1.5*iqr
lf
```

```python
np.var(data)
```

```python
np.std(data)
```

<h4>Skewness</h4>

```python
from scipy.stats import skew
skew(data)
```

<h4>Boxplot</h4>

```python
import seaborn as sns 
sns.boxplot(data)
```
