```python
import numpy as np 
import matplotlib.pyplot as plt
x=np.linspace(1,10,100)
```

```python
plt.plot(x,np.sin(x),c='g', linestyle='-.',label='sin')
plt.plot(x,np.cos(x),c='r', linestyle='--',label='cos')
plt.legend()
```

```python
plt.figure()
plt.subplot(1,2,1)
plt.plot(x,np.sin(x),c='g', linestyle='-.',label='sin')
plt.grid()
plt.legend()

plt.subplot(1,2,2)
plt.plot(x,np.cos(x),c='r', linestyle='--',label='cos')
plt.grid()
plt.legend()
```

<h4>pie chart : to understand the % distribution of data</h4>

```python
lang=['DS','DA','AIML','COMP','IT','AIDS','AI']
stud=[50,70,33,20,44,10,78]
explode=(0,0,0.1,0.1,0,0,0.1)
```

```python
plt.pie(stud,labels=lang,autopct='%1.1f%%',shadow=True,explode=explode,startangle=0)
plt.show()
```

```python
x=np.random.randint(1,10,size=10)   #random multiple
y=np.random.randint(1,10,size=10)
z=np.random.randint(1,10,size=10)
print(x,y,z)
```

```python
fig=plt.figure(figsize=(7,10))
ax=fig.add_subplot(projection='3d')

ax.scatter(x,y,z,c='r',label='X',alpha=1,marker='*')  
ax.scatter(x+0.2,y,z,c='g',label='Y',alpha=1,marker='D')
ax.scatter(x,y+0.2,z,c='b',label='Z',alpha=1,marker='x')  
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.legend()
plt.show()
```
