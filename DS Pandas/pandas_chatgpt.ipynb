```python
# Sex column में कितने male और female passengers हैं?

df['Sex'].value_counts()
```

```python
# Pclass column में कितने लोग हैं (1st, 2nd, 3rd class)?

df['Pclass'].value_counts()
```

```python
# 2. Selection / Filtering

# जिनकी age 18 से कम है, उनके नाम और age दिखाओ।abs

df[df['Age']<18][['Name','Age']]
```

```python
# Pclass 1 में कितने लोग थे और उनमें से कितने survived हुए?


pc=df[df['Pclass']==1 ]
len(pc)
```

```python
pc['Survived'].value_counts()
```

```python
# Fare 100 से ज़्यादा देने वाले passengers के नाम और fare दिखाओ।

df[df['Fare']>100][['Name','Fare']]
```

```python
# Female passengers जिनकी age 30 से ज़्यादा है, उनके नाम और Pclass दिखाओ।


df[(df['Sex']=='female')&(df['Age']>30)][['Name','Pclass']]
```

```python
# Male passengers जो survived हुए, उनके नाम और age दिखाओ।

df[(df['Sex']=='male')&(df['Survived']==1)][['Name','Age']]
```

```python
# Embarked column में "C" से embarked passengers के नाम दिखाओ।

df[df['Embarked']=='C']['Name']
```

```python
# SibSp > 2 वाले passengers के नाम और age दिखाओ।

df[df['SibSp']>2][['Name','Age','SibSp']]
```

```python
# 3. Aggregation / Statistics

# Average age कितनी है?

df['Age'].mean()
```

```python
# Maximum और minimum age कौन-कौन है?
df[df['Age']==df['Age'].max()][['Name','Age']]
```

```python
df[df['Age']==df['Age'].min()][['Name','Age']]
```

```python
# Average fare कितनी है?

df['Fare'].mean()
```
