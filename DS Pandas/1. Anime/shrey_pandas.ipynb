```python
# process for extracting episodes from the title
def ex_ep(txt):          #we want: (eps 10)
    check=False          #we don't want ..........().......
    d=""                 #we want to add(eps 100) in d
    for i in txt:        #we took i for every word from a txt
        if i=="(":       #when i comes here (
            
            check=True   #start to put data
                         
        if check==True:  #if there is True then start to add charcter
            d+=i         #add in data: eps 10
            
        if i==")":       #add last )
            check=False  #make it False
            return d     #add data in d
            
```

```python
# make a new column for episodes
df['Episodes']=df['Title'].apply(ex_ep)
```

```python
df['Episodes']=df['Episodes'].str.replace(r'\(|\)|\s*eps','',regex=True)
```

```python
df['Episodes']=df['Episodes'].astype(int)
```

```python
# process for Total_Time to extract from title
import re

def time_stamp(txt):
    # ) ke baad jo pattern month-year - month-year hai, usko match kare
    match = re.search(r'\)\s*([A-Za-z]{3}\s\d{4}\s*-\s*[A-Za-z]{3}\s\d{4})', txt)# ):\)  ' ':\s*  Apr:[A-Za-z]{3}  ' '2016:\s\d{4}  ' ':\s*  -  ' ':\s*
    if match:
        return match.group(1)   # matched date return karo
    return ''
```

```python
# make a new column for total_time declare
df['Time']=df['Title'].apply(time_stamp)
```

```python
# process for how many months took for anime
from dateutil import parser
import calendar

# Function to calculate months between two dates
def months_between(txt):
    start_str, end_str = txt.split(' - ')
    start_date = parser.parse(start_str)
    end_date = parser.parse(end_str)
    months = (end_date.year - start_date.year) * 12 + (end_date.month - start_date.month) + 1
    return months

# Apply function

```

```python
# make a new column for total_month to take
df['Total_Months'] = df['Time'].apply(months_between)
```
