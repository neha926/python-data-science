```python
एक 1D NumPy array बनाओ जिसमें 1 से 10 तक के numbers हों।

एक 3x3 का array बनाओ जिसमें सारे elements 0 हों।

एक random 4x4 matrix बनाओ और उसका transpose निकालो।

एक array बनाओ [10,20,30,40,50] और उसका mean, max और min निकालो।

एक 2D array (3x3) बनाओ और उसकी 2nd row print करो।
```

```python
import numpy as np
```

```python
# एक 1D NumPy array बनाओ जिसमें 1 से 10 तक के numbers हों।

np.arange(1,11)
```

```python
# एक 3x3 का array बनाओ जिसमें सारे elements 0 हों।

np.zeros([3,3])
```

```python
# एक random 4x4 matrix बनाओ और उसका transpose निकालो।

import random
s=np.random.randint(1,5,(4,4))
s
```

```python
s.T
```

```python
# एक array बनाओ [10,20,30,40,50] और उसका mean, max और min निकालो।

a=np.array([10,20,30,40,50]).max()
```

```python
a
```

```python
a=np.array([10,20,30,40,50]).mean()
```

```python
a
```

```python
a=np.array([10,20,30,40,50]).min()
```

```python
a
```
