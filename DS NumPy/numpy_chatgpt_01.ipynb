```python
# एक array बनाओ 1 से 50 तक और check करो कौन से numbers prime हैं। Output में सिर्फ prime numbers वाला array आना चाहिए।
import numpy as np
A = np.arange(1,51)
A
```

```python
def is_prime(n):
    if n<2:
        return False
    for i in range(2, int(np.sqrt(n))+1):
        if n%i==0:
            return False
    return True

prime_numbers = A[[is_prime(x) for x in A]]
print(prime_numbers)
```

```python
# एक 5x5 random matrix बनाओ और उसकी diagonal elements का sum निकालो (main diagonal और anti-diagonal दोनों)।

A=np.random.randint(1,10,(5,5))
A

```

```python
d=np.diag(A).sum()
d
```

```python
anti_diag_sum = np.diag(np.fliplr(A)).sum()
anti_diag_sum
```
