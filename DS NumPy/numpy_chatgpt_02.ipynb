```python
# Q10️⃣ (Super Hard)
# एक 10x10 matrix बनाओ जिसमें random integers 1–100 हों।
# उस matrix में:
# prime numbers को -1 से replace करो।
# multiples of 5 को 0 से replace करो।
# फिर:
# row-wise और column-wise sum निकालो।
# matrix में जो values positive हैं, उनका total sum निकालो।
# 💡 Hint:
# Prime numbers check करने के लिए vectorized method use करो।
# Masking + fancy indexing + logical operations बहुत useful होंगे।

Z=np.random.randint(1,101,(10,10))
Z

```

```python
def is_prime(n):
    if n<2:
        return False
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return False
    return True


pm=np.vectorize(is_prime)(Z)
Z[pm] = -1
      
        
```

```python
Z[Z%5==0]=0
Z
```

```python
r=Z.sum(axis=1)
r
```

```python
c=Z.sum(axis=0)
c
```

```python
p=Z[Z>0].sum()
p
```

```python
# Q11️⃣ (Masking + Fancy Index + Condition)
# एक 12x12 matrix बनाओ जिसमें random integers 1–150 हों।
# Matrix में:
# जो numbers prime और odd दोनों हैं, उन्हें -1 से replace करो।
# जो numbers multiples of 7 हैं, उन्हें 0 से replace करो।
# फिर:
# row-wise maximum निकालो।
# column-wise minimum निकालो।
```

```python
T=np.random.randint(1,151,(12,12))
T
```

```python
def is_prime(n):
    if n<2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n%i==0:
            return False
    return True


pm=np.vectorize(is_prime)(T)
odd=(T%2!=0)
T[pm]= -1
mask=pm+odd
T[mask]=-1
```

```python
T[T%7==0]=0
T
```

```python
r=T.max(axis=1)
r
```

```python
c=T.min(axis=0)
c
```

```python
# Q12️⃣ (Diagonal + Sum + Condition)
# एक 9x9 matrix बनाओ जिसमें 1–81 तक numbers हों।
# Main और secondary diagonals का sum निकालो।
# Diagonal में जितने भी even numbers हैं, उन्हें -2 से replace करो।
# फिर matrix का total sum निकालो।
```

```python
R=np.arange(1,82).reshape(9,9)
R
```

```python
main=np.diag(R)
main
```

```python
sec=np.diag(np.fliplr(R))
sec
```

```python
total = main.sum() + sec.sum()
total
```

```python
for i in range(9):
    if R[i, i] % 2 == 0:       # main diagonal
        R[i, i] = -2
    if R[i, 8 - i] % 2 == 0:   # secondary diagonal
        R[i, 8 - i] = -2
```

```python
R
```

```python
matrix_sum=R.sum()
matrix_sum
```
