```python
def list_item(a):
    n = []
    for i in a:
        if type(i) == list:
            for j in i:
                if type(j)==int:
                    n.append(j)
        else:
            if type(i) == int:
                n.append(i)
    return n

```

```python
print(list_item([1, 2.2, "Neha", True, 3, [10, 20, False]]))
```

```python
def sum(*a):
    total = 0
    p=1
    for i in a:
        total=total+i
        p=p*i
    return total,p
```

```python
sum(2,4,5,2,8,9)
```

```python
def mark_of_sub(**a):
    def total(markslist):
        return sum(markslist)
    markslist = []
    for sub,mark in a.items():
        markslist.append(mark)
    return total(markslist)
    
```

```python
mark_of_sub(math=98, science=88, english=92)
```

```python
def fib(n):
    a=0
    b=1
    for i in range(n):
        yield a
        a, b= b,a+b
        
```

```python
def fact(n):
    if n==0:
        return 1
    else: 
       return n*fact(n-1)
```

```python
fact(5)
```

```python
fact_num = lambda n: 1 if n==0 else n*fact_num(n-1)
fact_num(5)
```

```python
def sq(n):
    sq=[]
    for i in range(n):
        sq.append(i**2)
    return sq
```

```python
sq(5)
```
