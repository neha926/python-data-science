```python
from pathlib import Path 
```

```python
def read_fileandfolder():
    p=Path('.')   #current path
    items= list(p.rglob('*'))
    for i, items in enumerate(items):
        print(f"{i+1} : {items}")
```

```python
def createfile():
    try:
        read_fileandfolder()
        name=input("Enter File Name : ")
        p=Path(name)
        if not p.exists() and p.is_file():
            with open(p,"w") as fs:
                data=input("Which Content Do You Want To Add : ")
                fs.write(data)
            print("FILE CREATED")
        else:
            print("File Exist Already")
    except Exception as e:
        print(e)
        
        
```

```python
print("press 1 for Creating File : ")
print("press 2 for Reading File : ")
print("press 3 for Updating File : ")
print("press 4 for Deleting File : ")

check=int(input("Enter Your Choice : "))

if check==1:
    createfile()
```
