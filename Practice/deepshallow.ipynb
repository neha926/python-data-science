```python
a=[1,2,3,[5,6]]
```

<h4>General Copy : change in one, affect both</h4>

```python
b=a
```

```python
b[3][0]=33
```

```python
a
```

```python
b
```

<h4>Shallow Copy : Outer not modify original, but In inner - affect both</h4>

```python
b=a.copy()
```

```python
b[3][0]=55
```

```python
a
```

```python
b
```

<h4>Deep Copy : whatever you want to change inner or outer - doesn't affect original</h4>

```python
import copy
b=copy.deepcopy(a)
```

```python
b[0]=88
```

```python
a
```

```python
b
```

```python
b[3][0]=99
```

```python
a
```

```python
b
```
