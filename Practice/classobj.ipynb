```python
# Single Inheritance:


class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def show(self):
        print(f"Name of Person : {self.name}, age : {self.age}")   


class Student(Person):
        def __init__(self,name,age,roll,mark):
            super().__init__(name,age)
            self.roll=roll
            self.mark=mark
            
        def show(self):
            print(f"Name of Person:{self.name},  age:{self.age},  roll_no:{self.roll},  Marks:{self.mark}") 


p1=Person("Neha",23)
s1=Student("Yunsu",23,14,79)
s1.show()
```

```python
# multiple :  Father and Mother=>child

class Father:
    def __init__(self,sname):
        self.sname=sname

    def show(self):
        print(f"Surname Name: {self.sname}")

class Mother:
    def __init__(self,caste):
        self.caste=caste
         
    def show(self):
        print(f"Caste: {self.caste}")

class Child(Father,Mother):
    def __init__(self,sname,name,fname,caste):
        Father.__init__(self,sname)
        Mother.__init__(self,caste)
        self.name=name
        self.fname=fname
        

    def show(self):
          print(f"My Name is:{self.sname} {self.name} {self.fname},  And My Caste is :{self.caste}")

f=Father("Pawar")
m=Mother("Open")
c1=Child("Pawar","Neha","Tukaram" ,"96k Open")
c1.show()
```

```python
# Multi-level :  GrandFather=>Father=>Child



class GrandFather:
    def __init__(self,gfname,money):
        self.gfname=gfname
        self.money=money

    def show(self):
        print(f"Father Name:{self.gfname},   property:{self.money}")

class Father(GrandFather):
    def __init__(self,gfname,money,house):
        GrandFather.__init__(self,gfname,money)
        self.house=house
         
    def show(self):
        print(f"GrandFather Name:{self.gfname},   Property:{self.money},    house:{self.house}")

class Child(Father):
    def __init__(self,gfname,money,house,car):
        Father.__init__(self,gfname,money,house)
        self.car=car
        

    def show(self):
          print(f"GrandFather Name:{self.gfname},   Property:{self.money},    house:{self.house},   car:{self.car}")

gf=GrandFather("Sadashiv","7/12")
f=Father("Sadashiv","7/12","Building")
c1=Child("Sadashiv", "7/12" ,"Building", "BMW")
f.show()
```

```python
# Hierarchical:

class Mom:
    def __init__(self,mname):
        self.mname=mname

    def show(self):
          print(f"{self.mname} has 3 childern")


class d1(Mom):
    def __init__(self,mname,name1):
        Mom.__init__(self,mname)
        self.name1=name1

    def show(self):
          print(f"{self.mname} => {self.name1} is First Daughter")


class d2(Mom):
    def __init__(self,mname,name2):
        Mom.__init__(self,mname)
        self.name2=name2

    def show(self):
          print(f"{self.mname} => {self.name2} is Second Daughter")


class s1(Mom):
    def __init__(self,mname,name3):
        Mom.__init__(self,mname)
        self.name3=name3

    def show(self):
          print(f"{self.mname} => {self.name3} is Third Son")


m=Mom("Maya")
m.show()


d=d1("Maya","Amruta")
d.show()

d=d2("Maya","Neha")
d.show()
s=s1("Maya","Ruturaj")
s.show()
```

```python
# Hierarchical:

class GrandFather:
    def __init__(self,gname,**kwargs):
        super().__init__(**kwargs)
        self.gname=gname

    def show(self):
          print(f"{self.gname} has 2 childern")


class s1(GrandFather):
    def __init__(self,gname,name1,**kwargs):
        super().__init__(gname,**kwargs)
        self.name1=name1

    def show(self):
          print(f"{self.gname} has {self.name1} is First Daughter")


class s2(GrandFather):
    def __init__(self,gname,name2,**kwargs):
        super().__init__(gname,**kwargs)
        self.name2=name2

    def show(self):
          print(f"{self.gname} has {self.name2} is Second Daughter")


class GrandChild(s1,s2):
    def __init__(self,gname,name1,name2,name3):
        super().__init__(gname=gname,name1=name1,name2=name2)
        self.name3=name3

    def show(self):
          print(f"{self.gname} has 2 Daughter {self.name1} and {self.name2} and they have only one child {self.name3}")


g=GrandFather("Raj")
g.show()


s=s1("Raj","Amruta")
s.show()

s=s2("Raj","Neha")
s.show()

gc=GrandChild("Raj","Amruta","Neha","Ruturaj")
gc.show()
```
